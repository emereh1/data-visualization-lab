# Introduction

In this lab, I selected a dataset containing information about various reptile species, focusing on their classification across different taxonomic features. The primary objective of this analysis was to identify the target feature, Family, and explore how other features like Order, Subspecies, and Author relate to it. Using various visualizations, I aimed to determine which features are most predictive of the Family classification.


# Dataset Description

The dataset includes information on 12,440 reptile species or subspecies, with the following features:

ganzh: The scientific name of the reptile species or subspecies. Examples include Ablepharus alaicus and Ablepharus anatolicus. This is a textual feature, mainly used for identification.

Author: The author(s) responsible for the scientific classification of the reptile. Examples include "ELPATJEVSKY 1901" or "SCHMIDTLER 1997". This feature represents the historical classification source.

Subspecies: This feature refers to specific subspecies of the reptile, such as Ablepharus alaicus kucenkoi. If the subspecies is not available, it is filled with "Unknown".

Order: The taxonomic order to which the reptile belongs, such as Sauria. This represents the broader category in the classification hierarchy.

Family: The target feature of this analysis, representing the specific family to which the reptile belongs, such as Scincidae or Colubridae.

rdb_sp_id: A unique identifier for each entry in the dataset, which isn't relevant for analysis.

The dataset contains information about the classification of reptiles, focusing on their taxonomic relationships.

```python
# Importing necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Setting the file path for your dataset
file_path = r"C:\Users\emere\Desktop\Full Sail\DMV\DMV\Data_Visualization_And_Modeling-main\4_5_Data\reptile_checklist_2025_05.xlsx"

# Loading the dataset
dataset = pd.read_excel(file_path)

# Displaying the first few rows of the dataset
print(dataset.head())

# Cleaning the dataset
# Dropping columns with too many missing values
dataset_cleaned = dataset.drop(columns=['type_species', 'change'])

# Filling missing values in 'Subspecies' with a placeholder value 'Unknown'
dataset_cleaned['Subspecies'].fillna('Unknown', inplace=True)

# Checking for missing values in the cleaned dataset
print(dataset_cleaned.isnull().sum())

# Visualization: Distribution of Family
plt.figure(figsize=(10, 6))
sns.countplot(data=dataset_cleaned, x='Family', order=dataset_cleaned['Family'].value_counts().index)
plt.title('Distribution of Reptile Families')
plt.xticks(rotation=90)
plt.show()
```

# Distribution of Family

To understand the distribution of the Family feature, a count plot was created, showing the number of reptiles in each family. This helps us visualize how many species belong to each family and if any family is dominant or rare in the dataset.

Visualization: The plot revealed a diverse distribution across multiple families, with some families having a higher number of reptile species.

```python
# Visualization: Family by Order
plt.figure(figsize=(10, 6))
sns.countplot(data=dataset_cleaned, x='order', hue='Family')
plt.title('Distribution of Family by Order')
plt.xticks(rotation=90)
plt.show()
```

# Family by Order

Next, we examined how the Order feature relates to Family. A count plot was created to show the distribution of Family within each Order. This allowed us to observe how different orders correspond to particular families.

Visualization: The analysis indicated that certain Orders are associated with specific Families, suggesting a strong relationship between these two features.

```python
# Visualization: Family by Author
plt.figure(figsize=(10, 6))
sns.countplot(data=dataset_cleaned, x='Author', hue='Family')
plt.title('Distribution of Family by Author')
plt.xticks(rotation=90)
plt.show()
```

# Family by Author

We also explored the relationship between Author and Family. A count plot was generated to see if certain authors are consistently associated with specific families.

Visualization: The plot revealed that authors were distributed across multiple families, with no clear pattern showing that certain authors are tied to specific families.

```python
# Visualization: Family by Subspecies
plt.figure(figsize=(10, 6))
sns.countplot(data=dataset_cleaned, x='Subspecies', hue='Family')
plt.title('Distribution of Family by Subspecies')
plt.xticks(rotation=90)
plt.show()
```

# Family by Subspecies

Lastly, we explored how Subspecies relates to Family. This involved examining the distribution of Family for different Subspecies.

Visualization: The analysis showed that Subspecies did not significantly influence the Family classification, suggesting that the family classification is less dependent on subspecies.

# Results and Analysis

## Features Most Related to Family:

The Order feature shows the strongest relationship with Family. Different orders are strongly associated with certain families, making Order a good predictor of Family.

## Features Least Related to Family:

The Author and Subspecies features do not show a strong correlation with Family. These features appear less relevant in predicting the Family of a reptile.

# Big Picture Conclusion
We observe that 'Order' has a strong relationship with 'Family', while 'Author' and 'Subspecies' show weaker relationships.

From the visualizations and analysis, it is clear that the Order feature has the most significant influence on Family classification. This means that reptiles from specific orders tend to belong to particular families. In contrast, Author and Subspecies do not show strong relationships with the Family classification, suggesting they are not as useful for predicting a reptile's family.

Thus, for any future analysis or classification models, focusing on Order as a predictor for Family would be the most effective approach, while Author and Subspecies could be considered less important.
